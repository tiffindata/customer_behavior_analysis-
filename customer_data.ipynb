```python
import re

# Ensure column names are in the expected format before use
df.columns = [
    re.sub(r'_{2,}', '_', col.lower().replace(' ', '_').replace('-', '_'))
    for col in df.columns
]

df['add_ons_purchased'] = df.groupby('product_type')['add_ons_purchased'].transform(lambda x: x.fillna(x.mode()[0] if not x.mode().empty else 'None'))
df.isna().sum()
```

```python
# Convert column names to lowercase, replace spaces and hyphens with underscores,
# then replace any sequence of multiple underscores with a single underscore.
df.columns = [
    re.sub(r'_{2,}', '_', col.lower().replace(' ', '_').replace('-', '_'))
    for col in df.columns
]
# Display the new column names
df.columns
```

```python
#create a new column age_group
labels = ['Young Adult', 'Middle-aged', 'Senior']
df['age_group'] = pd.qcut(df['age'], q=3, labels=labels)
```
